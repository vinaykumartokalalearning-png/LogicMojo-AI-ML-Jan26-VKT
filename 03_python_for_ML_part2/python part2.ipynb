```python
## Main Application of set

A = {1,2,3,4,5}
B = {4,5,6,7,8}
print("Set A:", A)
print("Set B:", B)

print("A union B is:", A.union(B))
print("A intersection B is:", A.intersection(B))
B.issubset(A)
B.issuperset(A)
```

```python
print("=== Example 1: Building Vocabulary ===")

documents = [
    "machine learning is awesome",
    "deep learning is powerful",
    "machine learning and deep learning"
]

vocabulary = set()
for doc in documents:
    words = doc.lower().split()
    print("words are:", words)
    vocabulary.update(words)
    print("vocabulary is:", vocabulary)
    

## Question: Build a vocabulary set of unique words from these documents
```

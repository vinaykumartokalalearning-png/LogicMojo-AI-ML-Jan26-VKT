## What is a loop

```python
nums = [1, 2, 3, 4, 5] # multiple elements in the list

# loop we will use when some kind of operation with each and every element of the list
```

```python
# In Loop, we will do some operation with each and every element/argument

nums = [10,20,30,40]
for i in nums:           # loop through each element in nums. i (temporary variable mentioned) takes each value of nums list one by one
    print("inside the loop")
    print(i, i*2)
print("outside the loop")
```

```python
# Print only count/step of 2

nums = [10,20,30,40]
counter = 0
for i in nums:
    if (counter%2==0):
        print(i)
    counter = counter + 1
```

```python
# above one can simplified by using range() method

# range() method syntax is - range(start index, end index, value)

for i in range(0,10,2):
    print(i)
```

```python
# simplifed logic

nums = [10,20,30,40]
for i in range(0,len(nums),2):
    print(i, nums[i])
```

## Tuples

```python
# Tuples are similar to lists with minor difference. Tuples are ordered collection & immutable means can not be changes after the creation

point = (1,2,3) # define a tuple
point  # type is tuple
type(point)
```

```python
tuple_with_single_value = (3) # when you create a tuple with single value, it's madatory to use comma after the value else it will be treated as an integer
type(tuple_with_single_value) # it will considered as an integer, not tuple
```

```python
tuple_with_single_value = (3,) # always use comma when you create a tuple with single value
type(tuple_with_single_value) # it will be considered as a tuple since comma was passed after the value
```

```python
# empty tuple

empty = () # empty tuple
type(empty)
```

```python
# accessing the elements in Tuple. Tuples are ordered, so we can access via indexes

point = (1,2,3,4)
point[1] # index 1 is 2 from above tuple
```

```python
# unpacking the tuple by assigning multi-variables
point = (1,2,3,4)
x,y,z,w = point # unpacking tuple
print(x,y,z,w)
x,y = (5,6)
print(x,y,z,w) # seperate memory location
print(point) # no change in the actual tuple
```

```python
# most of the methods we used in Lists will work for Tuples

# indexing, slicing, length, step, count

list_count = [1,2,3,4,5,1,2,1]
list_count.count(1) # count takes only one argument

tuple_count = (1,2,5,4,1,3,2,1)
tuple_count.count(2) # count takes only one argument
```

```python
# tuple concatenation

point = (1,2,3,4)
tuple_concat = point + (5,6) # tuple concatenation

print(point, tuple_concat)
```

```python
point = (1,2,3,4) # ignore values while unpacking by using underscore (_)
a,_,c,_ = point # where you want to ignore, pass underscore (_)
print(a,c)
```

```python
# Tuple unpacking

coordinates = [(0,1), (2,3), (4,5)] # length is 3 since we count using comma (,)
x,y,z = coordinates
print(x,y,z)
```

```python
# Tuple unpacking in Loops

print("\n=== Unpacking in Loops ===")
coordinates = [(0,1), (2,3), (4,5)]
for x,y in coordinates:  # we are passing x,y variables to access & change the index values of easily
    print(x)
    print(y)
    print(f"Point: ({y}, {x})")
```

```python
# Tuple unpacking in Loops

print("\n=== Unpacking in Loops ===")
coordinates = [(0,1), (2,3), (4,5)]
for x in coordinates:  # we have access using index in each value/element if we are not passing 2 variables, here we are passing only 1 variable that is X
   print(x[0])
   print(x[1])
   print(x)
```

```python
# Unpacking ML example
# Simulating a batch of (features, labels) pairs, example is imagine label as age & features as age of multiple/different people
print("\n=== ML Example: Batch unpacking")

batch = [                 # To unpack we ahve 2 values only in the batch, not 4
    ([1.1,2.2,3.3], 0),
    ([2.1,2.6,5.6], 1),
    ([4.4,5.2,6.4], 0)
]

# Unpack & print featues and labels seperately. you deal with Features, Label everytime when you create an ML Model

for features,labels in batch:  # unpacking features & labels seperately
    print(f"Featues: {features}, Labels: {labels}")
    f1, f2, f3 = features
    L = labels
    print(f"f1: {f1}, f2: {f2}, f3: {f3} | Label: {L}")
```

```python
# Example , extract first and last values from below Tuple (Hint is use underscore to ignore the values while unpacking)

scores = (92, 85, 75, 95,97, 99, 88)
# first, _, _, _, last = scores
first, *_, last = scores  # Use *_ to ignore multiple values/elements in Tuple
print(f"First: {first}, Last: {last}")
```

```python
# Function arguments - data passed to a function are called arguments

#calling function & called function

def display(day,month,year):  # called function
    print("day:", day)
    print("month:", month)
    print("year",year)

def init():  # calling function
    display(2,2,2026)

init()
```

```python
# default argument values

def display(day,month,year = 2019):
    print("day:", day)
    print("month:", month)
    print("year",year)

display(2,2,2026)
```

```python
# keyword arguments
def display(day,month,year):
    print("day:", day)
    print("month:", month)
    print("year",year)
display(month = 2, day = 2, year = 2026)
```

```python
# Arbitary Arguments

def display(*data): # can be used when we don't know how many arguments to be passed
    for d in data:
        print(d)
display("Monday", 2, 2026)
```

```python
# Arbitary Arguments

def sum(*data):
    sum = 0
    for x in data:
        sum = x + sum
    print("sum is :", sum)

sum(10,20)
sum(1,2,3,4,5)
sum(50,40,60,20,10)
```

```python
# return statement - it terminates execution of called function & control returns to calling function

def fun():
    return 10
a = fun()
print(a)

def fun():
    return "abc"
a = fun()
print(a)

def fun():
    return 10, 20

a,b = fun()
print(a,b)

def sum(x,y):
    sum = x + y
    return sum
total = sum(10,20)
print(total)

# statment below return statement below won't be executed
```

```python
# Recursion - function calling itself, it shouldn't result in infinte loop aligning towards base case

def fun(n):
    print(n)
    if n == 1:
        return
    fun(n-1)

result = fun(5)
print(result)
```

```python
def factorial(n):
    if n <= 1:
        return n
    return n * factorial(n-1)

n = int(input())
if n >= 0:
    print(factorial(n))
```

```python
# Lambda Function - Anonymous function for short period of time, it can take any number arguments but only one expression is allowed

z = lambda x, y : x + y
z(1,2)
```

```python
# decorator function

def fun():
    print("Alan", end = " ")

def addgreeting(arg):
    def addingextrainfo():
        print("Hello", end = " ")
        arg()
        print("Good Morning!", end = " ")
    return addingextrainfo
new = addgreeting(fun)
new()  
```

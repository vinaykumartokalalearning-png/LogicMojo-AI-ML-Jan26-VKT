```python
# Main Application of Set

A = {1,2,3,4,5}
B = {4,5,6,7,8}
print("Set A:", A)
print("set B:", B)
```

```python
# Union

print("A.union(B) is", A.union(B))

# Intersection

print("A.intersection(B) is", A.intersection(B))

# Set difference

print(f"A - B = {A - B}")
print(F"B - A = {B - A}")

print(F"A.difference(B) = {A.difference(B)}")
print(F"B.difference(A) = {B.difference(A)}")

# Superset and Subset
A = {1,2,3,4,5}
C = {1,2,3}

print(f"C is subset of A: {C.issubset(A)}")
print(f"A is supersrt of C: {A.issuperset(A)}")
```

```python
## Question: Build a vocabulary set of unique words from these documents

print("=== Example 1: Building Vocabulary ===")

documents = [

    "machine learning is awesome",

    "deep learning is powerful",

    "machine learning and deep learning"

]

vocabulary = set()
for doc in documents:
    words = doc.lower().split()
    vocabulary.update(words)
print(f" Vocabulary is {vocabulary}")
```

```python
# Train data & Test Data Split validation

# To check data leakage b/w train & test data, use intersection to make sure no common data & no data leakage

train_ids = {1,2,3,4,5,6,7,8}
test_ids = {9,10,11,12}

overlap = train_ids.intersection(test_ids)

if not overlap:
    print("Data leakage not detected")
else:
    print("Data leakage detected")
```

```python
## Srach in O(1) ## find values of elements in large dataset

# Dictionary is a key & value pair here, it's unordered collection of key-value pair

#Keys must be unique & immutable, Vaues can be anything

employee_salary = {
    "Suvom": 5000,
    "Vinay": 4500,
    "Asakti": 6000,
    "Alice": 5500,
    "Bob": 4000
}

employee_salary["Vinay"]

# Grouping as below, with HR grouping. example we can group by HR, IT, Engineering Teams

employee_salary = { "HR":{
    "Suvom": 5000,
    "Vinay": 4500,
    "Asakti": 6000,
    "Alice": 5500,
    "Bob": 4000
}
}

employee_salary["HR"]["Vinay"]
```

```python
#slicing

numbers = [1,2,3,4,5,6]
# List[start index : end index] end index will be excluded
#slicing by default values List[:index], List[index:] default start will be 0 & end will be length of list

numbers[1:3] = [-2,-3]  # assignment
numbers

numbers[1:3] = [] # removing elements from the list
numbers

# copy = list_name[:] - copy of the ist syntax

numbers_copy = numbers[:] # it will copy numbers
print(numbers, numbers_copy)

# clear the whole list
numbers[:] = []
print(numbers)

#list methods

numbers = [1,2,3,4,5,6]
len(numbers)  # length of list

#appending items to the list - list.append(item)
numbers.append(7)  # append()  method takes exactly one argument
print(numbers)

# inserting items to the list - list.insert(index,item)
numbers.insert(0,0)
print(numbers)

#removing item from a list - list.remove(item)
numbers.remove(7)
print(numbers)

#removing item using del keyword
del numbers[2]
print(numbers)

#clear items from the list - list.clear()

numbers.clear()
print(numbers)

del numbers
# print(numbers)

```
